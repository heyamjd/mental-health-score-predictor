## 1. Load Data and Overview

```python
df = pd.read_csv('../data/raw/mental_health_data.csv')
print(f'Dataset Shape: {df.shape}')
df.head()
```

## 2. Missing Values Analysis

```python
missing = pd.DataFrame({
    'missing_count': df.isnull().sum(),
    'missing_pct': (df.isnull().sum() / len(df)) * 100
})
missing[missing['missing_count'] > 0]
```

## 3. Skewness Assessment

```python
skewness = df[NUMERIC_COLS + [TARGET_COL]].skew()
pd.DataFrame({'skewness': skewness})
```

## 4. Correlation Analysis

```python
corr = df[NUMERIC_COLS + [TARGET_COL]].corr()
plt.figure(figsize=(8, 6))
sns.heatmap(corr, annot=True, fmt='.2f', cmap='coolwarm', square=True)
plt.title('Correlation Heatmap')
plt.show()
```

## 5. Outlier Detection with IQR Method

```python
for col in NUMERIC_COLS:
    q25 = df[col].quantile(0.25)
    q75 = df[col].quantile(0.75)
    iqr = q75 - q25
    lower = q25 - 1.5 * iqr
    upper = q75 + 1.5 * iqr
    outliers = df[(df[col] < lower) | (df[col] > upper)]
    print(f'{col}: {len(outliers)} outliers outside [{lower:.2f}, {upper:.2f}]')
```

## 6. Target Distribution vs Stress Level

```python
plt.figure(figsize=(10, 5))
sns.boxplot(data=df.dropna(subset=['Stress_Level']), x='Stress_Level', y=TARGET_COL, order=['Low', 'Medium', 'High'])
plt.title('Mental Health Score Distribution Across Stress Levels')
plt.show()
```
